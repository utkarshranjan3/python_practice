```python
from collections import Counter, defaultdict
import sys
```

```python
# ==========================================
# 1. LISTS vs TUPLES & MEMORY EFFICIENCY
# ==========================================
my_list = [1, 2, 3, 4, 5]  # Mutable, dynamic array
my_tuple = (1, 2, 3, 4, 5)  # Immutable, fixed-size structure
```

```python
# Interview Tip: Tuples use less memory than lists because they are immutable
# and Python doesn't need to allocate extra space for future appends.
print(f"List size: {sys.getsizeof(my_list)} bytes")
print(f"Tuple size: {sys.getsizeof(my_tuple)} bytes")
```

```python
# ==========================================
# 2. DICTIONARIES & COLLECTIONS MODULE
# ==========================================
# Standard dict lookup is O(1) on average.
# defaultdict eliminates KeyError checks by providing a default factory.
dept_counts = defaultdict(int)
transactions = ["HR", "IT", "HR", "Finance", "IT", "IT"]
```

```python
for dept in transactions:
  # If 'dept' is missing, it defaults to 0 before incrementing
  dept_counts[dept] += 1

print("Department Counts (defaultdict):", dict(dept_counts))
```

```python
# Counter is a specialized dict subclass for counting hashable objects
# Heavily tested in data/coding interviews for frequency problems
word_counts = Counter(
    ["apple", "banana", "apple", "orange", "banana", "apple"]
)
print("Most Common Elements:", word_counts.most_common(1))  # [('apple', 3)]
```

```python
# ==========================================
# 3. SETS FOR O(1) MEMBERSHIP TESTING
# ==========================================
# Checking membership in a list is O(n). In a set, it is O(1) average.
valid_users = {"user_001", "user_002", "user_003"}  # Set lookup is O(1)
incoming_user = "user_002"

if incoming_user in valid_users:
  print(f"{incoming_user} is authorized.")
```

```python
import time
from functools import lru_cache

# ==========================================
# 1. LIST COMPREHENSIONS VS GENERATOR EXPRESSIONS
# ==========================================
# List Comprehension: Evaluates immediately and stores entire list in memory
squares_list = [x**2 for x in range(1000)]

# Generator Expression: Evaluates lazily (on-the-fly), using parentheses ()
# Interview Tip: Use generators when dealing with massive files or streams to prevent MemoryError.
squares_gen = (x**2 for x in range(1000))
print(
    "Generator object size (constant memory):"
    f" {sys.getsizeof(squares_gen)} bytes"
)
```

```python
# ==========================================
# 2. CUSTOM GENERATORS USING 'YIELD'
# ==========================================
def stream_large_dataset(file_rows, batch_size=2):
  """Simulates streaming data in chunks.

  Instead of loading all data into memory, 'yield' pauses the function
  execution and returns chunks sequentially.
  """
  for i in range(0, len(file_rows), batch_size):
    yield file_rows[i : i + batch_size]


mock_data = ["row1", "row2", "row3", "row4", "row5"]
for batch in stream_large_dataset(mock_data, batch_size=2):
  print("Processing batch:", batch)
```

```python
# ==========================================
# 3. DECORATORS & MEMOIZATION (@lru_cache)
# ==========================================
# Interviewers love asking about decorators. This one measures execution time.
def timing_decorator(func):
  def wrapper(*args, **kwargs):
    start_time = time.time()
    result = func(*args, **kwargs)
    end_time = time.time()
    print(
        f"Function '{func.__name__}' executed in"
        f" {end_time - start_time:.6f} seconds"
    )
    return result

  return wrapper


@timing_decorator
@lru_cache(maxsize=128)  # Caches results of expensive recursive/computed calls
def expensive_fibonacci(n):
  if n < 2:
    return n
  return expensive_fibonacci(n - 1) + expensive_fibonacci(n - 2)


print("Fibonacci(30) result:", expensive_fibonacci(30))
```
